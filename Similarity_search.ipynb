```python
import spacy
import numpy as np
from numpy.linalg import norm

# Load transformer-based spaCy model
nlp = spacy.load("en_core_web_trf")

# Example data
event_name_2 = ["love", "pizza", "hamburgers", "adore"]
des = "I adore hamburgers and enjoy eating pizza with friends."

# Process description once
doc_des = nlp(des)

# --- Helper function to get transformer embedding ---
def get_embedding(doc):
    # Works for all versions of spaCy-transformers
    if hasattr(doc._, "trf_last_hidden_state"):
        return doc._.trf_last_hidden_state.mean(axis=0)
    elif hasattr(doc._, "trf_data"):
        data = doc._.trf_data
        if hasattr(data, "model_output"):
            return data.model_output.last_hidden_state.mean(axis=0)
        elif hasattr(data, "tensors"):
            return data.tensors[-1].mean(axis=0)
    raise ValueError("No transformer embeddings found for this doc")

# --- Cosine similarity function ---
def cosine_similarity(v1, v2):
    return float(np.dot(v1, v2) / (norm(v1) * norm(v2)))

# --- Compute similarity for each word ---
desc_vec = get_embedding(doc_des)
for word in event_name_2:
    doc_word = nlp(word)
    word_vec = get_embedding(doc_word)
    sim = cosine_similarity(word_vec, desc_vec)
    print(f"{word:<15} → Similarity: {sim:.3f}")
```

```python
import spacy
import numpy as np
from numpy.linalg import norm

# Load transformer model
nlp = spacy.load("en_core_web_trf")

event_name_2 = ["love", "pizza", "hamburgers", "adore"]
des = "I adore hamburgers and enjoy eating pizza with friends."

# Process all texts together so the transformer runs once
docs = list(nlp.pipe([des] + event_name_2))
doc_des = docs[0]
event_docs = docs[1:]

# --- Helper: extract embedding safely ---
def get_embedding(doc):
    # Prefer new attribute if present
    if hasattr(doc._, "trf_last_hidden_state"):
        return doc._.trf_last_hidden_state.mean(axis=0)
    # Fall back to trf_data (older spaCy versions)
    if hasattr(doc._, "trf_data"):
        data = doc._.trf_data
        if hasattr(data, "model_output"):
            return data.model_output.last_hidden_state.mean(axis=0)
        if hasattr(data, "tensors"):
            return data.tensors[-1].mean(axis=0)
    raise ValueError("No transformer embeddings found in this Doc")

def cosine_similarity(v1, v2):
    return float(np.dot(v1, v2) / (norm(v1) * norm(v2)))

# --- Compute similarities ---
desc_vec = get_embedding(doc_des)
for word, doc in zip(event_name_2, event_docs):
    word_vec = get_embedding(doc)
    sim = cosine_similarity(word_vec, desc_vec)
    print(f"{word:<15} → Similarity: {sim:.3f}")
```

```python
import spacy
import numpy as np
from numpy.linalg import norm

# Load transformer model
nlp = spacy.load("en_core_web_trf")

# Example texts
texts = ["I hate pizza", "I adore hamburgers"]

# Run transformer only (for speed)
with nlp.select_pipes(enable=["transformer"]):
    docs = list(nlp.pipe(texts))

# --- Helper: extract transformer embedding safely ---
def get_embedding(doc):
    # New API: tensors is a list, each [tokens, hidden_size]
    if hasattr(doc._.trf_data, "tensors"):
        return doc._.trf_data.tensors[-1].mean(axis=0)
    # Fallback for older versions
    elif hasattr(doc._.trf_data, "last_hidden_state"):
        return doc._.trf_data.last_hidden_state.data.mean(axis=0)
    else:
        raise ValueError("Cannot find transformer embeddings in doc._.trf_data")

def cosine_similarity(v1, v2):
    return float(np.dot(v1, v2) / (norm(v1) * norm(v2)))

# Compute sentence similarity
vec1 = get_embedding(docs[0])
vec2 = get_embedding(docs[1])
similarity = cosine_similarity(vec1, vec2)

print(f"Cosine Similarity: {similarity:.3f}")
```

```python
import spacy
import numpy as np
from numpy.linalg import norm

# Load transformer model
nlp = spacy.load("en_core_web_trf")

# Example data
event_names = ["Telephone Detail Testing 1", "Telephone Detail Testing 2", "Fire outbreak", "Flood relief"]
description = "Heavy rainfall damaged several buildings and rescue teams were deployed."

# --- Helper: Get transformer embedding ---
def get_embedding(doc):
    # Compatible with spaCy-transformers >= 1.3
    if hasattr(doc._.trf_data, "tensors"):
        return doc._.trf_data.tensors[-1].mean(axis=0)
    else:
        raise ValueError("Transformer embeddings not found")

def cosine_similarity(v1, v2):
    return float(np.dot(v1, v2) / (norm(v1) * norm(v2)))

# --- Process all texts in one go for efficiency ---
docs = list(nlp.pipe([description] + event_names))
desc_vec = get_embedding(docs[0])
event_docs = docs[1:]

# --- Compute similarity for each event ---
similarities = []
for event, doc in zip(event_names, event_docs):
    event_vec = get_embedding(doc)
    score = cosine_similarity(desc_vec, event_vec)
    similarities.append((event, score))

# --- Sort by similarity ---
similarities.sort(key=lambda x: x[1], reverse=True)

# --- Print top match ---
predicted_event, top_score = similarities[0]
print(f"Predicted Event: {predicted_event}  (Score: {top_score:.3f})")
```

```python
import spacy
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load transformer-based spaCy model
nlp = spacy.load("en_core_web_trf")

# Your sentences
sentences1 = [
    "The new movie is awesome",
    "The cat sits outside",
    "A man is playing guitar",
]
sentences2 = [
    "The dog plays in the garden",
    "The new movie is so great",
    "A woman watches TV",
]

# Process all sentences efficiently
docs1 = list(nlp.pipe(sentences1))
docs2 = list(nlp.pipe(sentences2))

# Use the transformer output directly (mean pooling)
def get_sentence_vector(doc):
    # Each token has a vector from the transformer model
    # We take the mean across tokens to get a sentence embedding
    return doc._.trf_data.tensors[-1].mean(axis=0).flatten()

embeddings1 = np.array([get_sentence_vector(doc) for doc in docs1])
embeddings2 = np.array([get_sentence_vector(doc) for doc in docs2])

# Compute cosine similarity
similarities = cosine_similarity(embeddings1, embeddings2)

# Print results
for i, s1 in enumerate(sentences1):
    print(s1)
    for j, s2 in enumerate(sentences2):
        print(f" - {s2:<30}: {similarities[i][j]:.4f}")
```

```python
from transformers import AutoTokenizer, AutoModel
import torch
import torch.nn.functional as F

#Mean Pooling - Take attention mask into account for correct averaging
def mean_pooling(model_output, attention_mask):
    token_embeddings = model_output[0] #First element of model_output contains all token embeddings
    input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
    return torch.sum(token_embeddings * input_mask_expanded, 1) / torch.clamp(input_mask_expanded.sum(1), min=1e-9)


# Sentences we want sentence embeddings for
sentences = ['This is an example sentence', 'Each sentence is converted']

# Load model from HuggingFace Hub
tokenizer = AutoTokenizer.from_pretrained('sentence-transformers/all-MiniLM-L6-v2')
model = AutoModel.from_pretrained('sentence-transformers/all-MiniLM-L6-v2')

# Tokenize sentences
encoded_input = tokenizer(sentences, padding=True, truncation=True, return_tensors='pt')

# Compute token embeddings
with torch.no_grad():
    model_output = model(**encoded_input)

# Perform pooling
sentence_embeddings = mean_pooling(model_output, encoded_input['attention_mask'])

# Normalize embeddings
sentence_embeddings = F.normalize(sentence_embeddings, p=2, dim=1)

print("Sentence embeddings:")
print(sentence_embeddings)
```
